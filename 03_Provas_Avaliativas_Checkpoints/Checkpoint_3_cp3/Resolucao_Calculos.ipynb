### Formulação Matemática Formal
$$\text{Filtro DPI}: \text{Modbus TCP Function Code } 0x05 / 0x06 \xrightarrow{\text{Bloqueio se IP } \notin \text{ WhiteList}} \text{DROP}$$


### Passo 1: Separação de Zonas segundo a Norma ISA/IEC 62443
Divisão entre Nível 4 (Corporativo), Nível 3 (Supervisão/MES) e Nível 2 (Controle/CLPs).


### Passo 2: Zona Desmilitarizada Industrial (IDMZ)
Conexões entre IT e OT obrigatoriamente terminadas em proxies e servidores jump-box.


### Passo 3: Assinaturas de Snort / Suricata para Tráfego OT
Detecção de anomalias de varredura de portas Modbus e comandos de escrita em horário não-operacional.


### Passo 4: Prevenção de Conexões Reversas
Bloqueio estrito de qualquer tráfego originado da rede fabril em direção à internet pública.


### Passo 5: Conclusão Perimetral
Rede de automação imune a propagações laterais de ransomware corporativo.


```python
import numpy as np
import matplotlib.pyplot as plt

zonas = ['Nível 0/1 (Sensores)', 'Nível 2 (CLPs)', 'Nível 3 (SCADA)', 'IDMZ', 'Nível 4 (Corporativo)']
isolamento = [100, 95, 90, 85, 40]
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#210c0c')
ax.set_facecolor('#210c0c')
bars = ax.bar(zonas, isolamento, color='#ef4444', width=0.5)
ax.set_title('Nível de Isolamento por Camada IEC 62443 (%)', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_ylabel('Grau de Isolamento Crítico (%)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(axis='y', color='#4d1f1f', alpha=0.5)
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
