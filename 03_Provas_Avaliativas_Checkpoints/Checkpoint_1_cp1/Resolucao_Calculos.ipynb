### Formulação Matemática Formal
$$c \equiv m^e \pmod n, \quad m \equiv c^d \pmod n, \quad e \cdot d \equiv 1 \pmod{\phi(n)}$$


### Passo 1: Geração de Primos Fortes com Teste de Miller-Rabin
Seleção de p e q com probabilidade de primalidade > 99,999999%.


### Passo 2: Módulo RSA e Função Totiente
n = p \cdot q; \quad \phi(n) = (p - 1)(q - 1).


### Passo 3: Expoente Público e Inverso Modular
Adotar e = 65537; calcular d = e^{-1} \pmod{\phi(n)} via Algoritmo de Euclides Estendido.


### Passo 4: Assinatura Digital do Firmware do Robô
S = H(firmware)^d \pmod n, verificável publicamente por qualquer nó via S^e \pmod n.


### Passo 5: Conclusão Criptográfica
Proteção absoluta contra injeção de firmware malicioso nos controladores industriais.


```python
import numpy as np
import matplotlib.pyplot as plt

tamanhos_chave = np.array([1024, 2048, 4096])
tempo_quebra_anos = np.array([0.01, 1e9, 1e25])
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#210c0c')
ax.set_facecolor('#210c0c')
ax.plot(tamanhos_chave, tempo_quebra_anos, 'o-', color='#ef4444', lw=2.5)
ax.set_yscale('log')
ax.set_title('Resistência a Ataques de Fatoração (RSA)', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Tamanho da Chave RSA (Bits)', color='#94a3b8')
ax.set_ylabel('Tempo Estimado para Quebra (Anos)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#4d1f1f', alpha=0.5)
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
