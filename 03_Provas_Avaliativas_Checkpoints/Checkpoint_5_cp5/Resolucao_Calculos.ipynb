### Formulação Matemática Formal
$$\text{Regra SIEM}: \text{count}(\text{event.type: failed\_login}) > 5 \text{ em } 60\text{s} \implies \text{Alerta SOC Severidade Alta}$$


### Passo 1: Ingestão de Logs com Beats e Logstash
Coleta centralizada de logs de autenticação Linux, auditoria de CLPs e tráfego de rede.


### Passo 2: Normalização no Elastic Common Schema (ECS)
Padronização de campos source.ip, user.name e event.outcome.


### Passo 3: Correlação Automatizada de Alertas
Geração de incidentes correlacionando falha de login seguida de elevação de privilégio sudo.


### Passo 4: Playbooks de Resposta Automatizada (SOAR)
Isolamento automático da porta do switch afetada via integração de API de rede.


### Passo 5: Conclusão Operacional
Tempo médio de resposta a incidentes (MTTR) reduzido para menos de 90 segundos.


```python
import numpy as np
import matplotlib.pyplot as plt

minutos = np.arange(1, 11)
eventos_normais = np.random.poisson(25, 10)
eventos_normais[5] = 180  # ataque
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#210c0c')
ax.set_facecolor('#210c0c')
ax.plot(minutos, eventos_normais, 'o-', color='#ef4444', lw=2.5, label='Eventos de Falha de Login / Minuto')
ax.axhline(60, color='#eab308', linestyle='--', label='Limiar de Alarme SIEM')
ax.set_title('Detecção de Força Bruta por Correlação de Logs no SIEM', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Janela de Tempo (Minutos)', color='#94a3b8')
ax.set_ylabel('Eventos por Minuto', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#4d1f1f', alpha=0.5)
ax.legend(facecolor='#331515', edgecolor='#ef4444', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
