### Formulação Matemática Formal
$$\text{Maturidade NIST} = \frac{1}{5} \sum (\text{Identificar} + \text{Proteger} + \text{Detectar} + \text{Responder} + \text{Recuperar}) \ge 4,2$$


### Passo 1: Mapeamento de Dados Sensíveis e Inventário de Ativos (Identify)
Classificação de informações e avaliação de impacto à privacidade (DPIA).


### Passo 2: Medidas Protetivas e Políticas de Menor Privilégio (Protect)
Criptografia de dados em trânsito e em repouso com controle estrito de acessos.


### Passo 3: Monitoramento Contínuo e Threat Hunting (Detect)
Detecção contínua de ameaças com equipe de SOC dedicada 24/7.


### Passo 4: Plano de Resposta a Incidentes Cibernéticos (Respond)
Procedimentos formais de contenção e comunicação à ANPD em até 48 horas.


### Passo 5: Conclusão de Governança
Auditoria independente aprovando maturidade nível 4 (Gerenciado e Mensurável).


```python
import numpy as np
import matplotlib.pyplot as plt

funcoes = ['Identificar', 'Proteger', 'Detectar', 'Responder', 'Recuperar']
maturidade = [4.5, 4.8, 4.2, 4.0, 4.3]
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#210c0c')
ax.set_facecolor('#210c0c')
bars = ax.bar(funcoes, maturidade, color='#ef4444', width=0.5)
ax.set_ylim(0, 5.0)
ax.set_title('Nível de Maturidade por Função do NIST CSF (Escala 1 a 5)', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_ylabel('Escore de Maturidade', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(axis='y', color='#4d1f1f', alpha=0.5)
for bar in bars:
    yval = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2.0, yval + 0.1, f'{yval}', ha='center', va='bottom', color='#ffffff', fontweight='bold')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
